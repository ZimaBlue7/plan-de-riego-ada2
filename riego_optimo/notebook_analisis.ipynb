# Análisis de Rendimiento - Riego Óptimo
Este notebook utiliza los módulos del proyecto para generar fincas aleatorias, ejecutar los algoritmos y visualizar las métricas de rendimiento. Todo importando directamente desde los `.py` ya creados.

```python
# Importar los módulos del proyecto
from generador import generar_finca_aleatoria
from analisis import comparar_tres_algoritmos, benchmark_escalabilidad, resumen_tabla

import pandas as pd
import matplotlib.pyplot as plt
```

## 1. Comparación en una finca individual

```python
# Generar una finca aleatoria de 8 tablones
finca = generar_finca_aleatoria(n=8, seed=42)
print(f"Finca generada: {finca.n} tablones\n")

# Ejecutar los 3 algoritmos y obtener métricas
metricas = comparar_tres_algoritmos(finca)

# Mostrar el resumen en formato tabla ASCII
print(resumen_tabla(metricas))
```

## 2. Benchmark de Escalabilidad (Tiempo vs N)

```python
# Ejecutar benchmark para diferentes tamaños de finca (n)
# Omitimos valores muy altos para que la Fuerza Bruta no tarde demasiado
tamanos_n = [3, 4, 5, 6, 7, 8, 9, 10]
resultados_bench = benchmark_escalabilidad(tamanos_n, repeticiones=2)

# Convertir a DataFrame de Pandas para mejor visualización
df = pd.DataFrame([
    {'Algoritmo': m.algoritmo, 'N': m.n, 'Costo': m.costo, 'Tiempo (ms)': m.tiempo_ms, 'Memoria (bytes)': m.memoria_bytes}
    for m in resultados_bench
])
display(df)
```

## 3. Gráfica de Escalabilidad

```python
# Graficar los tiempos de ejecución
fig, ax = plt.subplots(figsize=(10, 6))

for algo in df['Algoritmo'].unique():
    datos_algo = df[df['Algoritmo'] == algo]
    ax.plot(datos_algo['N'], datos_algo['Tiempo (ms)'], marker='o', label=algo)

ax.set_xlabel('Número de tablones (n)')
ax.set_ylabel('Tiempo de ejecución (ms)')
ax.set_title('Escalabilidad de Algoritmos (Tiempo vs N)')
ax.legend()
ax.grid(True)
# plt.yscale('log') # Descomentar para ver en escala logarítmica
plt.show()
```
